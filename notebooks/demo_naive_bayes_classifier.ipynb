# Naive Bayes Classifier Demo
This notebook shows how to train and compare the three Naive Bayes variants (Gaussian, Multinomial, Bernoulli) on the Iris dataset using the modular code in this repository.

## Bayes' theorem in simple terms
Bayes' theorem updates how likely a hypothesis is after seeing new evidence. If $P(A)$ is the prior belief and $P(B|A)$ describes how likely the evidence is under that belief, then the posterior is given by $P(A|B) = rac{P(B|A)P(A)}{P(B)}$.

**Why 'naive'?** Naive Bayes assumes that features are conditionally independent given the class. This is rarely true, but the simplification keeps the model fast and often performs well in practice when the assumption is approximately correct.

## When to use each variant
- **GaussianNB**: continuous numeric features (e.g., measurements).
- **MultinomialNB**: counts or frequencies (e.g., bag-of-words).
- **BernoulliNB**: binary indicators (e.g., word present/absent).

```python
from app.data import load_iris_dataset
from app.preprocess import split_data, transform_features_for_variants
from app.model import get_models
from app.evaluate import evaluate_models

# 1. Load data
X, y = load_iris_dataset()
print(X.head())

# 2. Train/test split and feature preparation
split = split_data(X, y)
features = transform_features_for_variants(split, scale_gaussian=False)

# 3. Train and evaluate
models = get_models()
features_dict = {
    'gaussian': features.gaussian,
    'multinomial': features.multinomial,
    'bernoulli': features.bernoulli,
}
metrics = evaluate_models(models, features_dict, split.y_train, split.y_test)

metrics
```

## Quick comparison
You can tweak `scale_gaussian=True` to standardize inputs for GaussianNB. For text data, replace the dataset with a bag-of-words representation and keep using MultinomialNB or BernoulliNB.
